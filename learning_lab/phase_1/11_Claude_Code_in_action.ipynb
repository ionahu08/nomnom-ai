## 1. What is Claude Code
### 1.1 Introduction to code assistant
#### 1.1.1 what is code assistant
A code assistant is a system that uses **language models** to handle complex programming tasks like reading code, fixing bugs, and editing files. It's more than a code generator — it's a model + tooling system that can interact with your actual environment.

**Key takeaways**:
- Code assistants use language models to complete different tasks
- Language models need **tools** to handle most real-world programming work (LLMs alone can only generate text)
- Not all language models use tools with the same skill level
- **Claude's strong tool use** enables better security, customization, and longevity in Claude Code


#### 1.1.2 How code assistant works
When you give a code assistant a task (e.g., "fix this bug"), it follows the same 3 steps a human developer would:

1. **Gather context** — Understand what the error means, which files are involved, what the codebase looks like
2. **Formulate a plan** — Decide what to change, what to test, in what order
3. **Take action** — Actually edit files, run commands, verify the fix

**The catch**: Steps 1 and 3 require interacting with the **outside world** — reading files, running commands, fetching docs. But language models by themselves can **only process and return text** — they can't read files or run commands directly. This gap is solved by **tool use**.


#### 1.1.3 Main Challenges + Solution + how the solution works
##### The challenge
LLMs can only output text. They can't actually read files or run commands. Ask a raw LLM to read `main.go` and it'll just tell you it can't.

##### The solution: Tool Use
The code assistant injects instructions into the request that teach the model how to **request actions through text**. The model says *"I want to read this file"* via a special format, and the assistant (running outside the model) actually performs the action.

##### How it works — end-to-end flow
1. You ask: *"What's in main.go?"*
2. Code assistant **adds tool instructions** to your message before sending it to the model
3. Model responds: `ReadFile: main.go` (formatted text request)
4. Code assistant **parses the request** and actually reads the file from disk
5. Code assistant sends the file contents **back to the model**
6. Model produces the final answer based on real file contents

The LLM never actually touches your filesystem — it just generates carefully formatted text requests, and the surrounding code assistant does the real work.

##### Why Claude's tool use matters
Claude (Opus / Sonnet / Haiku) is particularly strong at understanding what tools do and combining them. This gives Claude Code three concrete advantages:

| Benefit | What it means |
|---|---|
| **Tackles harder tasks** | Claude chains multiple tools and adapts to tools it has never seen before |
| **Extensible platform** | You can add new tools and Claude will figure out how to use them — no retraining needed |
| **Better security** | Claude navigates codebases without pre-indexing, meaning your entire codebase doesn't have to be sent to external servers |


##### Putting it together: what Claude Code actually is

> **Claude Code = strong LLMs (Opus / Sonnet / Haiku) + a comprehensive set of built-in tools**

Claude Code ships with built-in tools for common dev tasks — reading files, writing code, running commands, managing directories. The real power isn't any single tool, but how Claude **intelligently combines** them to tackle complex, multi-step problems.

---

#### 💡 One-line takeaway
> A code assistant = LLM + tool use. The LLM generates text-formatted action requests, the assistant runs them in the real world, and feeds results back. Claude Code works well because Claude is exceptionally good at the "tool use" half of this equation.

### 2.3 Adding context
#### 2.3.1 /init command 
After analyzing your code, Claude creates a summary and writes it to a **CLAUDE.md** file. When Claude asks for permission to create this file, you can either hit Enter to approve each write operation, or press Shift+Tab to let Claude write files freely throughout your session.

#### 2.3.2 Claude.md file
##### 2.3.2.1 Two purposes of Claude.md file
- Guides Claude through your codebase, pointing out important commands, architecture, and coding style
- Allows you to give Claude specific or custom directions

##### 2.3.2.2 Three types of Claude.md file locations
- **./CLAUDE.md** (Project Level)

    Generated with /init, committed to source control, shared with other engineers

- **./CLAUDE.local.md** (Local Level)

    Not shared with other engineers, contains personal instructions and customizations for Claude

- **~/.claude/CLAUDE.md** (Machine Level)

     Used with all projects on your machine, contains instructions that you want Claude to follow on all projects


##### 2.3.2.3 Adding Custom Instructions into Claude.md
You can customize how Claude behaves by adding instructions to your CLAUDE.md file. For example, if Claude is adding too many comments to code, you can address this by updating the file. Edit CLAUDE.md directly in your editor, or run /memory(put "/memory" at the beginning of a chat message) inside Claude Code to open the file. Add an instruction like Use comments sparingly. Only comment complex code.


##### 2.3.2.4 Adding Custom Instructions into Claude.md --> Referencing Files in CLAUDE.md
You can also mention files directly in your CLAUDE.md file using the same @ syntax. This is particularly useful for files that are relevant to many aspects of your project.

For example, if you have a database schema file that defines your data structure, you might add this to your CLAUDE.md:

The database schema is defined in the @prisma/schema.prisma file. Reference it anytime you need to understand the structure of data stored in the database.



### 2.4 Making changes
#### 2.4.1 Two ways to improve Claude Code's intelligence
You can combine both modes for tasks that require both breadth and depth.

- Way 1:Planning mode (Breadth-first): /plan or by pressing Shift + Tab **twice**

    Tip: when reviewing the plan, you can press Ctrl+G to open it in your text editor. You can precise edits before approving the plan, and Claude will see the final version you submit.

- Way 2: Extended thinking (Depth-first)

    1) If you want to signal to Claude that it should do extra thinking on a single prompt, use the keyword ultrathink in your prompt. This signals to Claude that it should reason more on this turn, but doesn't adjust the session's effort level.

    2) If you want to see Claude's reasoning process, press Ctrl+O to expand the actual reasoning steps.

    3) You can control is how Claude reasons through a problem by setting an effort level. Run /effort to see your current level and adjust it: low is faster and cheaper, max reasons longest on hard problems. The default depends on your model and plan — /effort shows you what yours is. 


### 2.5 Controlling context
Several techniques you can use to control the flow of your conversation and help Claude stay on track when working with Claude code on complex tasks.

#### 2.5.1 Interrupting Claude with Escape
#### 2.5.2 Combining Escape with Memories
When Claude makes the same mistake repeatedly across different conversations, you can:

Press Escape to stop the current response
Run /memory (or edit CLAUDE.md directly) to add a note about the correct approach
Continue the conversation with the corrected information
This prevents Claude from making the same error in future conversations on your project.

#### 2.5.3 /rewind or pres ESC twice
This shows you all the messages you've sent, allowing you to jump back to an earlier point and continue from there. This technique helps you:

- Maintain valuable context (like Claude's understanding of your codebase)
- Remove distracting or irrelevant conversation history
- Keep Claude focused on the current task

#### 2.5.4 Context Management Commands
##### 2.5.4.1 /compact
The /compact command summarizes your entire conversation history while preserving the key information Claude has learned. This is ideal when the conversation has become long but contains important context.

##### 2.5.4.2 /clear
The /clear command starts a new conversation with fresh context. This is most useful when:

You're switching to a completely different, unrelated task
The current conversation context might confuse Claude for the new task
You want to start over without any previous context
You can still go back to the previous conversation later with /resume. The /clear command does not remove the conversation from your session history.



### 2.6 Custom Commands
### 2.6.1 Creating Custom Commands
To create a custom command, you need to set up a specific folder structure in your project:

- Find the .claude folder in your project directory
- Create a new directory called commands inside it
- Create a new markdown file with your desired command name (like audit.md)

The filename becomes your command name - so audit.md creates the /audit command.

### 2.6.2 Commands with Arguments with Arguements
Custom commands can accept arguments using the $ARGUMENTS placeholder. This makes them much more flexible and reusable.

## 3. Hooks and SDK
### 3.1 Introduction to the hooks
#### 3.1.1 What is hook and two most common types of hooks
Hooks in Claude Code allow you to intercept and control tool calls before or after they execute. This gives you fine-grained control over what Claude can and cannot do in your development environment.

Two of the most common hook types are below (you'll see other hooks in a later lesson):

- PreToolUse hooks - Run before a tool is called
- PostToolUse hooks - Run after a tool is called

#### 3.1.2 Hook configuration
Hooks are defined in Claude settings files. You can add them to:

- Global - ~/.claude/settings.json (affects all projects)
- Project - .claude/settings.json (shared with team)
- Project (not committed) - .claude/settings.local.json (personal settings)

#### 3.1.3 Some common ways to use hook
Here are some common ways to use hooks:

- Code formatting - Automatically format files after Claude edits them
- Testing - Run tests automatically when files are changed
- Access control - Block Claude from reading or editing specific files
- Code quality - Run linters or type checkers and provide feedback to Claude
- Logging - Track what files Claude accesses or modifies
- Validation - Check naming conventions or coding standards



### 3.2 Defining a hook + Implementing a hook
Creating a hook involves four main steps:
1. Decide on a PreToolUse or PostToolUse hook - PreToolUse hooks can prevent tool calls from executing, while PostToolUse hooks run after the tool has already been used
2. Determine which type of tool calls you want to watch for - You need to specify exactly which tools should trigger your hook
3. Write a command that will receive the tool call - This command gets JSON data about the proposed tool call via standard input
4. If needed, command should provide feedback to Claude - Your command's exit code tells Claude whether to allow or block the operation


### 3.3 Useful Hooks
1. TypeScript Type Checking Hook

One of the most useful hooks addresses a fundamental problem: when Claude modifies a function signature, it often doesn't update all the places where that function is called throughout your project.

2. Query Duplication Prevention Hook

In larger projects with many database queries, Claude sometimes creates duplicate functionality instead of reusing existing code. This is especially problematic when you give Claude complex, multi-step tasks that include database operations as just one component.
