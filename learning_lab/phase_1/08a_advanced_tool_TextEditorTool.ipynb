# Text Editor Tool — Notes

## Core Concept

A **predefined tool** built into Claude. Anthropic provides the schema; you write the implementation.

## My Summary (in plain words)

Sometimes we want Claude to operate on local files — common actions include viewing, editing, creating, undoing, and so on. But Claude has no access to our local file system and no idea what our files look like. It can only "issue instructions" telling us what to do; the actual execution is on us.

Without the Text Editor Tool, we'd have to write a separate schema + function for every single operation — tedious and easy to miss cases. Since file operations are so common and generic, **Anthropic has predefined a complete set of schemas for us**. This isn't several independent tools — it's **one tool (`str_replace_editor`) with multiple commands** (view, create, str_replace, insert, undo_edit) baked in.

But Anthropic can't access our computer for us, so **the actual execution functions still have to be written by us**.

To use it, we just drop a **minimal stub** (`{"type": "text_editor_20250124", "name": "str_replace_editor"}`) into the API call. Anthropic's servers automatically expand it into the full predefined schema so Claude can see it. When Claude decides to call the tool, it returns a `tool_use` request with a `command` field. **Our dispatcher code routes that command to the corresponding function**, executes it, and sends the result back to Claude.


## Key Distinction

- **Custom tool**: you write schema + implementation
- **Text Editor tool**: Anthropic provides schema, you write **only the implementation**

> Claude knows how to *request* file operations, but you must write the code that actually performs them (create files, read directories, replace text, etc.).

## What It Can Do

- View file / directory contents (with optional line ranges)
- Replace text in a file
- Create new files
- Insert text at a specific line
- Undo recent edits

## How to Use It

Include a **minimal schema stub** in your request. Claude expands it into the full spec internally:

​```python
tools = [{
    "type": "text_editor_20250124",  # version string is model-dependent
    "name": "str_replace_editor",
}]
​```

**Version note**: different Claude models require different `type` strings. Check the official docs for the right one:
https://docs.anthropic.com/en/docs/agents-and-tools/tool-use/text-editor-tool

## When to Use It

Only valuable when:

- Your app needs to **programmatically** edit files
- You're working in an environment without an AI code editor
- You want to embed file-editing capabilities into your own Claude-powered app

## One-Line Takeaway

Lets you replicate an "AI-powered code editor" inside your own application — Claude Code itself is essentially an extended version of this tool.
